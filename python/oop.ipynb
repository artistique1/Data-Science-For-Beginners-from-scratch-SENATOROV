```python
# Видео 3. Атрибуты класса и функции класса 

# Класс создает свое пространство имен - это специальный словарь, в котором в виде ключ-значение хранятся имена переменных и функции с их значениями. Определение свойств класса делается через присваивание переменным некоторых значений. 
# Например:
# class Person:
#     name = 'Ivan'  # мы создали переменную и присвоили ей значение. нюанс в том, что эти переменные должны быть объявлены внутри класса. В соответствии с этой записью python создаст свойство (атрибут, поле) name для класса Person. 
# Все свойства можно посмотреть с помощью функции 
# dir(Person)

# Состояние любых объектов хранится в словарях dict
Person.__dict__  # это и есть собственно пространство имен классов и экземпляров классов (не экземпляров классов). Возвращает объект mappingproxy, который можно трактовать как read-only словарь: он ведет себя полностью как словарь, за исключением того, что нельзя изменять значения ключей непосредственно через обращение к этому словарю, и нельзя ничего добавить или удалить.
# Код ниже вызовет ошибку TypeError:
# Person.__dict__["name"] = "vfghfgh"
# Мы можем влиять на словарь dict или только через dot-нотацию, или через специальные встроенные функции getattr(), setattr() и delattr().

# Обращение к переменной name (свойству класса) делается через точку:
Person.name
# Класс создает новое пространство имен, в котором мы можем объявлять новые переменные и функции, которые становятся его аттрибутами. Так как python динамический язык, он позволяет нам добавлять новые свойства классам уже после их определения прям на лету. 
Person.age = 23
Person.__dict__  # тут мы видим динамически созданный атрибут age класса Person

getattr(Person, 'age')  # используется для чтения значений свойств класса

setattr(Person, 'dob', '12.03.2000')  # позволяет устанавливать новое свойство со значением или можно изменить значение уже имеющегося атрибута
Person.__dict__

delattr(Person, 'dob')  # удаляем свойство
Person.__dict__

# Так же как и переменные мы можем объявлять функции внутри класса
class NewPerson:
    name = 'Kseniya'
    
    def hello():
        print("Hello")
        
# NewPerson.hello()
print(NewPerson.__dict__) # если мы вызовем словарь dict, то увидим, что hello это функция
```

```python
# Видео 4. Классы как callable-объекты, экземпляры классов. 

# Классы - это т.н. callable-объекты, то есть вызываемый объект, как функция.
# когда мы создаем класс с помощью ключевого слова class, python автоматически добавляет к нему некоторые свойства и поведение. И в том числе добавляет к нему метод __call__, который делает все классы callable-объектами. 
# Если вызвать класс, он возвращает свой экземпляр (объект)
print(Person.__dict__)

p1 = Person()
# p1  # в ответ получаем объект экземпляра класса Person
p2 = Person()

# Если сравнить их id, то они будут разными, то есть это разные объекты
print(id(p1), id(p2))

# Мы создаем разные экземпляры классов, чтобы хранить разные значения одних и тех же их свойств. Проблема в том, что свойства класса глобальны для всех объектов. Например, свойство name будет одинаково для обоих экземпляров класса Person, несмотря на то, что это 2 разных объекта. 
p1.name
p2.name

# А id свойства name будет одинаковым. Более того, это будет тот же самый объект, определенный в классе Person.
print(id(p1.name), id(p2.name), id(Person.name))

# Если обратиться к словарям экземпляров классов, мы увидим, что они пусты. Пространства имен каждого экземпляра класса пусты, поэтому python пошел искать их выше - в их родительском классе. Такое поведение оправдано, потому что такой подход позволяет экономить память и ресурсы на создание атрибутов и хранение их значений. На этом механизме основано наследование и полиморфизм в python. 
print(p1.__dict__, p2.__dict__)

# Свойства класса как бы создают поведение по умолчанию и создают состояния объектов по умолчанию. Именно поэтому нам нужно создавать экземпляры классов, так как у каждого экземпляра есть свое собственное пространство имен, свой словарь dict,. И поэтому каждый экземпляр может хранить свое уникальное состояние, то есть словари dict у каждого экземпляра класса и самого класса будут отличаться.
p1.name = "Ксения"
p2.name = "Юра"
print(p1.__dict__, p2.__dict__)
p2.age = 23
print(p1.__dict__, p2.__dict__)
# Таким образом мы динамически создали новые свойства (атрибуты) обоим экземплярам класса Person и присвоили им новые значения, которые были записаны в локальные словари этих экземпляров. При этом словарь dict самого класса Person будет неизменным.

# Если попытаться обратиться к свойству age экземпляра класса p1, получаем ошибку AttributeError, поскольку объект p1 не имеет атрибута age. 
# p1.age
# Сначала python попытается найти атрибут age в экземпляре p1, если он его там не находит, то обращается в родительский класс - если его и там нет, то выходит эта ошибка.

# Создадим новые экземпляры классов Person и переопределим значение атрибута name в самом классе Person. Атрибут name изменится во всех экземплярах.
p1 = Person()
p2 = Person()
Person.name = 'Oleg'
print(p1.name, p2.name, Person.name)
```

```python
# Видео 5. Функции классов и методы экземпляров.

class Person:
    def hello():
        print("Hello")
        
# Person.hello
# Выведется <function __main__.Person.hello()> - hello это именно объект функции.  

# А теперь создадим экземпляр класса и посмотрим на объект hello с его точки зрения
p = Person()
p.hello
# <bound method Person.hello of <__main__.Person object at 0x000001D84A6361E0>> - в данном случае это bounded-метод (связанный метод) и дальше идет указание на основной класс __main__.Person
hex(id(p))
# id экземпляра класса Person будет иметь тот же самый адрес, что и класс Person - 0x1d84b5f0740
# Другими словами это говорит о том, что атрибут hello у переменной p, которая является экземпляром класса Person, является связанным методом с объектом hello у класса Person. И их поведение будет отличаться друг от друга
# Person.hello()  # Hello
# p.hello()  # TypeError: Person.hello() takes 0 positional arguments but 1 was given

# Посмотрим на типы объектов атрибута hello у экземпляра класса p и самого класса Person
type(p.hello)  # method
type(Person.hello)  # function
# То есть функции классов, при создании экземпляров этого класса, трансформируются в методы и связываются с только что созданным экземпляром класса.  Это будут разные объекты с разными id:
id(Person.hello)  # 2028490217472
id(p.hello)  # 2028490404864
# Плюс у них будут разные атрибуты:
dir(Person.hello)
dir(p.hello)  # у экземпляра появился атрибут __self__

# Методы и функции существуют не сами по себе, они нужны для обработки состояний, которые были сохранены в экземпляре класса. А пространство имен класса и экземпляра друг от друга изолированы. И чтобы функция класса работала как нам надо, нам нужно передать в нее экземпляр этого класса, в пространстве имен которого сохранено его состояние. Другой возможности дотянуться до сохраненного значения у класса нет, только получив экземпляр этого класса. Именно для этого python и связывает с методом экземпляра сам экземпляр, чтобы класс мог обратиться к своему экземпляру и получить доступ к его пространству имен. То есть под капотом python вызывает функцию, которая определена в классе таким образом:
# Person.hello(p)  # здесь получаем исключение, так как не указали что функция hello принимает в себя какие-то аргументы, но под капотом вызов метода выглядит именно так

print(hex(id(p)), p.hello.__self__)
p.hello.__func__
# При вызове метода и передаче в него аргумента под капотом происходит следующее:
# p.hello.__func__(hello.__self__, *args)
# Поэтому в самом классе нужно предусмотреть момент, что при вызове метода из экземпляра туда будет передаваться сам экземпляр, то есть необходимо учесть этот неявны дополнительный первый аргумент для методов экземпляра класса. 
# По соглашению первый обязательный параметр методов экземпляров классов называется self.
class NewPerson:
    def hello(self):
        print("Hello")
        
p1 = NewPerson()
p1.hello()
```

```python
# Видео 6. Инициализация экземпляров, __init__ метод.

from tkinter import NE


class Person:
    def create(self):
        self.name = "Ivan"
        
    def display(self):
        print(self.name)
        
p = Person()
# p.display()  # AttributeError
p.__dict__  # Это происходит из-за того, что локальный словарь dict у экземпляра класса на этапе создания пустой и имя name в родительском классе тоже не определено. 
p.create()
p.__dict__

# это все неудобно. Поэтому есть специальный метод init для задания свойств экземплярам классов и присваивании им первоначальных значений при их создании.Этот метод вызывается автоматически при создании экземпляра класса. 
class NewPerson:
    def __init__(self, name):
        self.name = name
        
    def display(self):
        print(self.name)
        
p1 = NewPerson("Ivan")
p1.name
p1.__dict__

# Init метод это не совсем конструктор класса. Создание экземпляров классов состоит из 2 этапов: 1) метод __new__() создает сам экземпляр - это и есть конструктор класса 2) метод __init__ - инициализирующий метод, инициализирует экземпляр свойствами и их значениями. 
```
