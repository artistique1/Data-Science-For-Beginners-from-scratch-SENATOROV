```python
# 5.  

from sys import stdin

lines: list = stdin.readlines()
palindromes: list = []
for line in lines:
    x = line[:-1].split()
    for i in x:
        if i.lower() == i[::-1].lower() and i not in palindromes:
            palindromes.append(i)
for i in sorted(palindromes):
    print(i)
```

```python
# 6.  

gost = {
    'А': 'A', 'Б': 'B', 'В': 'V',
    'Г': 'G', 'Д': 'D', 'Е': 'E',
    'Ё': 'E', 'Ж': 'ZH', 'З': 'Z',
    'И': 'I', 'Й': 'I', 'К': 'K',
    'Л': 'L', 'М': 'M', 'Н': 'N',
    'О': 'O', 'П': 'P', 'Р': 'R',
    'С': 'S', 'Т': 'T', 'У': 'U',
    'Ф': 'F', 'Х': 'KH', 'Ц': 'TC',
    'Ч': 'CH', 'Ш': 'SH', 'Щ': 'SHCH',
    'Ы': 'Y', 'Э': 'E', 'Ю': 'IU',
    'Я': 'IA', 'Ь': '', 'Ъ': '',
}

data, translit_data = '', ''
with open("cyrillic.txt", encoding="UTF-8") as file_in:
    for line in file_in:
        data += line

for i in data:
    if i.upper() in gost:
        translit_data += gost[i.upper()].lower().capitalize() if i == i.upper() else gost[i.upper()].lower()
    else:
        translit_data += i

with open("transliteration.txt", "w", encoding="UTF-8") as file_out:
    print(translit_data, file=file_out)
```

```python
# 8.  

input_filenames = input(), input()
output_filename = input()
words = [set(), set()]
for i in range(len(input_filenames)):
    with open(input_filenames[i], 'r') as file_in:
        for line in file_in:
            words[i].update({x for x in line.split()})
with open(output_filename, 'w') as output_filename:
    for word in sorted(words[0].symmetric_difference(words[1])):
        print(word, file=output_filename)
```

```python
# 11.  

import json

first_file, second_file = input(), input()
numbers: list = []
with open(first_file) as f:
    for line in f:
        numbers.extend([int(x) for x in line.split()])
data = {
        "count": len(numbers),
        "positive_count": len(list(filter(lambda x: x > 0, numbers))),
        "min": min(numbers),
        "max": max(numbers),
        "sum": sum(numbers),
        "average": round(sum(numbers) / len(numbers), 2)
}
with open(second_file, "w", encoding="UTF-8") as g:
    json.dump(data, g, ensure_ascii=False, indent=4)
```

```python
# 12.  

from sys import stdin

files: list = [name.strip() for name in stdin]
data: list = []
with open(files[0]) as f:
    for line in f:
        data.append(line.split())
D = dict(zip([1, 2, 3], ['>', '<', '==']))
for n in range(1, len(files)):
    with open(files[n], 'w') as f:
        for line in data:
            for i in line:
                if eval('len(list(filter(lambda x: not int(x) % 2, i)))'
                        + D[n] +
                        'len(list(filter(lambda x: int(x) % 2, i)))'):
                    print(i, end=' ', file=f)
            print('\n', end='', file=f)
```

```python
# 13.  

from sys import stdin
import json

data: list = [line.strip() for line in stdin]
print(data)
print({line.split(' == ')[0]: line.split(' == ')[1] for line in data[1:]})
D: dict = {line.split(' == ')[0]: line.split(' == ')[1] for line in data[1:]}
with open(data[0], 'r', encoding="UTF-8") as f:
    records = json.load(f)
records.update(D)
if len(data) > 1:
    with open(data[0], 'w', encoding="UTF-8") as f:
        json.dump(records, f, ensure_ascii=False, indent=4, sort_keys=True)
```

```python
# 14.  

import json

first_file, second_file = input(), input()
with open(first_file, 'r') as f:
    data_1 = json.load(f)
with open(second_file, 'r') as f:
    data_2 = json.load(f)
data_1: dict = {i["name"]: {k: v for k, v in i.items() if k != "name"} for i in data_1}
for d in data_2:
    if d["name"] in data_1:
        for key in d:
            if (key not in data_1[d["name"]] or d[key] > data_1[d["name"]][key]) and key != "name":
                data_1[d["name"]][key] = d[key]
    else:
        data_1[d["name"]] = {k: v for k, v in d.items() if k != "name"}
with open(first_file, 'w') as g:
    json.dump(data_1, g, ensure_ascii=False, indent=4)
```

```python
# 15.  

from sys import stdin
import json

answers: list = [i.strip() for i in stdin]
with open("scoring.json", "r") as file_in:
    data = json.load(file_in)
data: list = [{y["pattern"]: x["points"] // len(x["tests"]) for y in x["tests"]} for x in data]
result: int = 0
counter: int = 0
for i in data:
    for j in range(counter, len(i) + counter):
        if answers[j] in i:
            result += i[answers[j]]
        counter += 1
print(result)
```

```python
# 16.  

from sys import stdin

data: list = [line.strip() for line in stdin]
query: str = data[0].lower()
D: dict = {}
for file in data[1:]:
    with open(file, "r", encoding="UTF-8") as f:
        D[file] = f.read()
        if query in ' '.join(D[file].replace('&nbsp;', ' ').lower().split()):
            print(file)
        else:
            del D[file]
if not D:
    print('404. Not Found')
```

```python
# 18.  

import os

size = os.path.getsize(input())
if size > 1024**3 - 1:
    size = int(size / 1024**3) + 1
    postfix = 'ГБ'
elif size > 1024**2 - 1:
    size = int(size / 1024**2) + 1
    postfix = 'МБ'
elif size > 1023:
    size = int(size / 1024) + 1
    postfix = 'КБ'
else:
    postfix = 'Б'
print(str(size) + postfix)
```

```python
# 19.  

cipher: int = int(input())
a: str = "abcdefghijklmnopqrstuvwxyz"
with open("public.txt", "r", encoding="UTF-8") as file_in:
    data = file_in.read()
data_out = [a[(a.find(i.lower()) + cipher) % len(a)] if i.lower() in a else i for i in data]
for ind, letter in enumerate(data):
    if letter.isupper():
        data_out[ind] = data_out[ind].upper()
with open("private.txt", "w") as file_out:
    print(''.join(data_out), file=file_out)
```
