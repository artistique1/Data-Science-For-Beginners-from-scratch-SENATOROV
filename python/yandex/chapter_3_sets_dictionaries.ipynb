```python
# 3.  

n = int(input())
objects = set(input().split())
for _ in range(n - 1):
    new_objects = set(input().split())
    objects = objects.union(new_objects)
for obj in objects:
    print(obj)
```

```python
# 4.  

n, m = int(input()), int(input())
man, ovs = set(), set()
for _ in range(n):
    man.add(input())
for _ in range(m):
    ovs.add(input())
print(len(man & ovs) if man & ovs else "Таких нет")
```

```python
# 5.  

n, m = int(input()), int(input())
one_porridge = set()
for _ in range(n + m):
    if (child := input()) in one_porridge:
        one_porridge.remove(child)
    else:
        one_porridge.add(child)
print(len(one_porridge) if one_porridge else "Таких нет")
```

```python
# 6.  

n, m = int(input()), int(input())
one_porridge = set()
for _ in range(n + m):
    if (child := input()) in one_porridge:
        one_porridge.remove(child)
    else:
        one_porridge.add(child)
if one_porridge:
    for child in sorted(one_porridge):
        print(child)
else:
    print("Таких нет")
```

```python
# 8.  

n: int = int(input())
favorite_porridge = {}
for _ in range(n):
    pair: list = input().split()
    favorite_porridge[pair[0]] = pair[1:]
porridge: str = input()
counter: int = 0
for child in sorted(favorite_porridge):
    if porridge in favorite_porridge[child]:
        print(child)
        counter += 1
if not counter:
    print("Таких нет")
```

```python
# 9.  

all_objects = dict()
str_number = 0
while (text := input()) != "":
    elements = text.split()
    for element in elements:
        if element not in all_objects:
            all_objects[element] = 1
        else:
            all_objects[element] += 1
for element in all_objects:
    print(f"{element} {all_objects[element]}")
```

```python
# 12.  

names: list = []
n: int = int(input())
for _ in range(n):
    names.append(input())
counter: int = 0
unique_names = set(names)
for name in sorted(unique_names):
    if names.count(name) > 1:
        print(f"{name} - {names.count(name)}")
        counter += 1
if not counter:
    print("Однофамильцев нет")
```

```python
# 13.  

n: int = int(input())
menu: set = set()
old_menu: set = set()
for _ in range(n):
    menu.add(input())
m: int = int(input())

for _ in range(m):
    for _ in range(int(input())):
        old_menu.add(input())

new_menu: set = menu - old_menu
if new_menu:
    for dish in sorted(new_menu):
        print(dish)
else:
    print("Готовить нечего")
```

```python
# 14.  

goods: set = set()
for _ in range(int(input())):
    goods.add(input())
recipies: dict = {}
for _ in range(int(input())):
    recipies[dish := input()] = set()
    for _ in range(int(input())):
        recipies[dish].add(input())
counter: int = 0
for dish in sorted(recipies):
    if not recipies[dish] - goods:
        print(dish)
        counter += 1
if not counter:
    print("Готовить нечего")
```

```python
# 16.  

objects: set = set()
while area := input().split():
    for i, near in enumerate(area):
        if near == "зайка":
            if i == 0:
                objects.add(area[i + 1])
            elif i == len(area) - 1:
                objects.add(area[i - 1])
            else:
                objects.add(area[i + 1])
                objects.add(area[i - 1])
for obj in sorted(objects):
    print(obj)
```

```python
# 17.  

friends = {}
while x := input().split():
    if x[0] not in friends:
        friends[x[0]] = {x[1]}
    else:
        friends[x[0]].add(x[1])
    if x[1] not in friends:
        friends[x[1]] = {x[0]}
    else:
        friends[x[1]].add(x[0])
friends_2 = dict.fromkeys(friends, set())
for friend in friends:
    for n in friends[friend]:
        friends_2[friend] = friends_2[friend].union(friends[n])
    friends_2[friend].discard(friend)
    for z in friends[friend]:
        friends_2[friend].discard(z)
data = []
for friend in friends_2:
    data.append(f'{friend}: {", ".join(sorted(friends_2[friend]))}')
data.sort()
for string in data:
    print(string)
```

```python
# 18.  

points: dict = {}
n: int = int(input())
for _ in range(n):
    coordinates: list = input().split()
    if not (point := f'{coordinates[0][:-1]}-{coordinates[1][:-1]}') in points:
        points[point] = 1
    else:
        points[point] += 1
print(max(points.values()))
```

```python
# 19.  

data = []
for _ in range(int(input())):
    k = list(map(lambda x: x.rstrip(','), input().split()))
    data.extend(set(k[1:]))
data = sorted(toy for toy in data if data.count(toy) == 1)
for kid in data:
    print(kid)
```

```python
# 20.  

data = sorted(map(int, input().split('; ')))
result = dict.fromkeys(data)
for i in data:
    for j in data:
        a, b = i, j
        while b:
            a, b = b, a % b
        if a == 1:
            if result[i]:
                result[i].add(j)
            else:
                result[i] = {j}
for number in result:
    if result[number]:
        print(f'{number} - {", ".join(map(str, sorted(result[number])))}')
```
