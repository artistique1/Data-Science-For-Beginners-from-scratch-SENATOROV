```python
# 5.  

n: int = int(input())
counter, tmp = 0, 0
for _ in range(n):
    while (x := input()) != 'ВСЁ':
        if x == 'зайка':
            tmp += 1
    if tmp:
        counter += 1
        tmp = 0
print(counter)
```

```python
# 8.  

n: int = int(input())
winner: str = ""
number: int = 0
for _ in range(n):
    name: str = input()
    summa: int = sum(list(map(int, input())))
    number = max(number, summa)
    if number == summa:
        winner = name
print(winner)
```

```python
# 11.  

n: int = int(input())
counter: int = 0
for _ in range(n):
    j: int = 2
    if (num := int(input())) > 1:
        counter += 1
        if num == 2:
            counter += 1
        while num % j:
            if j > num ** 0.5:
                break
            j += 1
        else:
            counter -= 1
print(counter)
```

```python
# 14.  

n, m = int(input()), int(input())
width = len(str(n * m))
for i in range(1, n + 1):
    if i % 2:
        for j in range(m * (i - 1) + 1, m * i + 1):
            if j == m * i:
                print(str(j).rjust(width, ' '))
            else:
                print(str(j).rjust(width, ' '), end=' ')
    else:
        for j in range(m * i, m * (i - 1), -1):
            if j == m * (i - 1) + 1:
                print(str(j).rjust(width, ' '))
            else:
                print(str(j).rjust(width, ' '), end=' ')
```

```python
# 15.  

n, m = int(input()), int(input())
width = len(str(n * m))
for i in range(1, n + 1):
    for j in range(m):
        if not j % 2 and j != m - 1:
            print(str(i + 2 * n * (j // 2)).rjust(width, ' '), end=' ')
        elif not j % 2 and j == m - 1:
            print(str(i + 2 * n * (j // 2)).rjust(width, ' '))
        elif j % 2 and j != m - 1:
            print(str(2 * n * (j // 2 + 1) - (i - 1)).rjust(width, ' '), end=' ')
        else:
            print(str(2 * n * (j // 2 + 1) - (i - 1)).rjust(width, ' '))
```

```python
# 16.  

a, b = int(input()), int(input())
for i in range(1, a + 1):
    for j in range(1, a + 1):
        if j != a:
            print(f'{(str(i * j) + " " if b % 2 else str(i * j)).center(b)}|', end='')
        else:
            print(f'{(str(i * j) + " " if b % 2 else str(i * j)).center(b)}')
    if i * j != a * a:
        print(f'{"-" * ((b + 1) * a - 1)}')
```

```python
# 18.  

n, lengths = '', [0]
for j in range(1, (x := int(input())) + 1):
    n += str(j) + ' '
    if j in (sum(range(i)) for i in range(j + 2)):
        lengths.append(len(n) - 1)
        n = ''
lengths.append(len(n) - 1)
d = 1
for z in range(1, x + 1):
    if z - 1 in (sum(range(i)) for i in range(z + 2)):
        print(f"{' ' * ((max(lengths) - lengths[d]) // 2)}{z}", end=' ' if z != 1 else '\n')
        d += 1
    else:
        print(z, end='\n' if z in (sum(range(i)) for i in range(z + 2)) else ' ')
```

```python
# 19.  

n: int = int(input())
for i in range(n):
    for j in range(n):
        d = str(min(i, j, n - i - 1, n - j - 1) + 1)
        print(d.rjust(len(str((n + 1) // 2)), ' '), end=' ' if j < n - 1 else '\n')
```

```python
# 20.  

def convert_from_base_to_base(num, from_base=10, to_base=10):
    n = int(num, from_base) if isinstance(num, str) else num
    alphabet = "0123456789ABCDEFGHIJKLMNOPQRSTUVWXYZ"
    result = ""
    while n > 0:
        n, m = divmod(n, to_base)
        result += alphabet[m]
    return result[::-1]


number = input()
max_benefit, best_base = 0, 0
for base in range(2, 11):
    if (digits_sum := sum(map(int, convert_from_base_to_base(number, 10, base)))) > max_benefit:
        best_base = base
        max_benefit = digits_sum
print(best_base)
```
