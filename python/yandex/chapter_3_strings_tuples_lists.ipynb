```python
# 10.  

text = ""
while (input_string := input()) != "ФИНИШ":
    text += input_string.lower().replace(" ", "")

letters = set(text)
char_count = {}
for letter in letters:
    char_count[letter] = text.count(letter)
max_count = max(char_count.values())
most_common_chars = [char for char, count in char_count.items() if count == max_count]
most_common_char_sorted = sorted(most_common_chars)[0]
print(most_common_char_sorted)
```

```python
# 15.  

input_numbers = list(map(int, input().split()))


def gcd(a, b):
    while b:
        a, b = b, a % b
    return a


def find_gcd_of_list(numbers):
    if not numbers:
        return 0
    current_gcd = numbers[0]
    for num in numbers[1:]:
        current_gcd = gcd(current_gcd, num)
        if current_gcd == 1:
            break
    return current_gcd


result = find_gcd_of_list(input_numbers)
print(result)
```

```python
# 19.  

expression: list = input().split()
stek = []
for i in expression:
    if i.isdigit():
        stek.append(int(i))
    else:
        a = stek.pop()
        b = stek.pop()
        if i == "-":
            stek.append(b - a)
        elif i == "+":
            stek.append(b + a)
        elif i == "*":
            stek.append(b * a)
print(stek[0])
```

```python
# 20.  

def factorial(n):
    return n * factorial(n - 1) if n > 1 else 1


expression: list = input().split()
stek = []
for i in expression:
    if i.isdigit():
        stek.append(int(i))
    else:
        if i in ["-", "+", "*", "/"]:
            a = stek.pop()
            b = stek.pop()
            if i == "-":
                stek.append(b - a)
            elif i == "+":
                stek.append(b + a)
            elif i == "*":
                stek.append(b * a)
            elif i == "/":
                stek.append(b // a)
        elif i == "~":
            stek[-1] = -stek[-1]
        elif i == "@":
            stek.append(stek.pop(-3))
        elif i == "!":
            stek[-1] = factorial(stek[-1])
        elif i == "#":
            stek.append(stek[-1])
print(stek[0])
```
