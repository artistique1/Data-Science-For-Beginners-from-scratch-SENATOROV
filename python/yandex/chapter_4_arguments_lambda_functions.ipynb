```python
# 2.  

def make_matrix(size, value=0):
    if isinstance(size, int):
        return [[value for _ in range(size)] for _ in range(size)]
    else:
        return [[value for _ in range(size[0])] for _ in range(size[1])]
```

```python
# 3.  

def gcd(*args):
    numbers = list(args)
    while len(numbers) > 1:
        while numbers[1]:
            numbers[0], numbers[1] = numbers[1], numbers[0] % numbers[1]
        numbers.pop(1)
    return numbers[0]
```

```python
# 5.  

def to_string(*args, **kwargs):
    return kwargs.get("sep", " ").join([str(i) for i in args]) + kwargs.get("end", "\n")
```

```python
# 10.  

def secret_replace(text, **kwargs):
    result: str = ""
    kwargs = {key: (value, 0) for key, value in kwargs.items()}
    for i in text:
        if i in kwargs:
            result += kwargs[i][0][kwargs[i][1] % len(kwargs[i][0])]
            kwargs[i] = kwargs[i][0], kwargs[i][1] + 1
        else:
            result += i
    return result
```
