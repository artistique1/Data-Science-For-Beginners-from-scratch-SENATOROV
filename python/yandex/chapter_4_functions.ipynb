```python
# 2.  

def gcd(a, b):
    while b:
        a, b = b, a % b
    return a
```

```python
# 7.  

def can_eat(horse, other) -> bool:
    return abs(horse[0] - other[0]) + abs(horse[1] - other[1]) == 3
```

```python
# 8.  

def is_palindrome(obj) -> bool:
    if isinstance(obj, int):
        obj = str(obj)
    return obj == obj[::-1]
```

```python
# 9.  

def is_prime(number):
    for i in range(2, int(number**0.5) + 1):
        if not number % i:
            return False
    return number != 1
```

```python
# 10.  

def merge(tuple_1, tuple_2) -> tuple:
    numbers = list(tuple_1) + list(tuple_2)
    length = len(numbers)
    for i in range(length):
        for j in range(0, length - i - 1):
            if numbers[j] > numbers[j + 1]:
                numbers[j], numbers[j + 1] = numbers[j + 1], numbers[j]
    return tuple(numbers)
```
