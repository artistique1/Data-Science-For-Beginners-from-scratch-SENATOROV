Функция `np.where()`

```python
# предположим, что у нас есть массив с оценками учеников на контрольной работе,
# и нам нужно преобразовать их массив вида "Сдал" / "Не сдал"
scores = np.array([5, 2, 4, 3, 2, 4])

# вначале создадим массив из логических значений
condition = scores > 2
condition
```

```python
# теперь возьмем два списка с соответствующими категориями
# такой же длины как и исходный массив
pass_list = ['Pass'] * len(scores)
fail_list = ['Fail'] * len(scores)

# посмотрим на один из списков
pass_list
```

```python
# передадим функции np.where() логические значения и два списка
np.where(condition, pass_list, fail_list)
```

```python
# если не передавать массивы с категориями, можно узнать, например,
# индексы двоечников ИЛИ отличников
indices = np.where((scores < 3) | (scores > 4))
indices
```

```python
# можно и так
indices = np.where(np.logical_or(scores < 3, scores > 4))
indices
```

```python
# такой результат можно также использовать как фильтр исходного массива
scores[indices]
```

Функция активации ReLU

```python
# можно использовать встроенную функцию np.maximum()
def ReLU(x):
  return np.maximum(0, x)
```

```python
# создадим последовательность точек на оси x
x = np.linspace(-10, 10, 1000)

# передадим их в функцию ReLU
y = np.maximum(0, x)

# задидим размер и сетку и выведем график функции
plt.figure(figsize = (8, 6))
plt.grid()
plt.plot(x, y)
plt.title('ReLU')
plt.show()
```

```python
# можно написать и собственную функцию
def ReLU_arr(x):
  return x * (x > 0)
```

Данные

```python
# возьмем три наблюдения
# вектор данных будет иметь размерность 3 x 1
X = np.array([[185], [180], [175]])
X.shape
```

Веса и смещения

```python
# первая матрица весов w1 и w2 с размерностью 1 x 2
wh = np.array([[0.4, 0.2]])

# смещения b1 и b2 (достаточно одномерного массива)
bh = np.array([2, 8])

# вторая матрица весов 2 x 2
# w3, w5
# w4, w6
wo = np.array([[0.7, 0.18], [0.6, 0.47]])

# и смещения b3 и b4
bo = np.array([5, 9])
```

Прямое распространение

```python
# объявим функцию прямого распространения
def forward(X):

  # вначале рассчитаем скрытый слой
  hidden = ReLU(np.dot(X, wh) + bh)

  # затем выходной
  output = ReLU(np.dot(hidden, wo) + bo)

  # и выведем результат
  return output
```

```python
# и вызовем ее
forward(X)
```

#### Решение систем линейных уравнений

```python
# создадим матрицу коэффициентов при неизвестных
A = np.array([[4, -2, 3],
              [1, 3, -4],
              [3, 1, 2]])

# и вектор-столбец свободных коэффициентов
b = np.array([[1],
              [-7],
              [5]])
```

### Статистика в Numpy

```python
# сгенерируем данные о росте 1000 человек и поместим их в список
np.random.seed(42)
height = list(np.round(np.random.normal(180, 10, 1000)))
print(height)
```

#### Меры среднего

Среднее арифметическое

```python
# найдем среднее арифметическое
np.mean(height)
```

Медиана

```python
# медиану
np.median(height)
```

```python
# то есть 50-й процентиль
np.percentile(height, 50)
```

Средневзвешенное значение

```python
# поместим оценки и соответствующие веса в списки
scores = [3, 4, 5]
weights = [0.5, 0.4, 0.1]
```

```python
# передадим их в функцию np.average() и найдем средневзвешенное значение
np.average(scores, weights = weights)
```

```python
# для вычисления вручную, распакуем оценки
linalg, calculus, history = scores

# и веса в отдельные переменные
w1, w2, w3 = weights

# перемножим оценки и веса и сложим получившиеся произведения
w1 * linalg + w2 * calculus + w3 * history
```
